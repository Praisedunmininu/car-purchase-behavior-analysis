This project explores customer data to understand the factors influencing car purchase amounts.




```python

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

## Load Dataset
We load the dataset containing customer demographic and financial information.

```python
df = pd.read_csv("car_purchasing.csv", encoding='latin1')
df.head()
```

## Initial Data Exploration
We explore the structure and summary of the dataset.

```python
df.info()
df.describe()
df.shape
df.columns
```

## Gender Analysis
We analyze how gender affects car purchase behavior.

```python
df["gender"].value_counts()
```

```python
gender_spending = df.groupby("gender")["car purchase amount"].mean().reset_index()
gender_spending
```

```python
gender_spending.plot(x="gender",y="car purchase amount",kind='bar')
plt.title("Average car purchase by gender")
plt.show()
```

*Insight:*  
Gender shows only a slight difference in both purchase count and spending. It is not a strong factor influencing car purchase.

## Net Worth Analysis
We group customers based on net worth to observe spending patterns

```python
net_worth_bins = [0,299824,426750,557324,1000000]
net_worth_labels = ["low","medium","high","very_high"]

df["net_worth_group"] = pd.cut(df["net worth"], bins=net_worth_bins, labels=net_worth_labels)

net_worth_analysis = df.groupby("net_worth_group")["car purchase amount"].mean().reset_index()
net_worth_analysis
```

```python
net_worth_analysis.plot(x="net_worth_group", y="car purchase amount", kind='bar')
plt.title("Average Car Purchase by Net Worth")
plt.show()
```

*Insight:*  
Customers with higher net worth spend significantly more. This is the strongest factor influencing car purchase.

## Age Analysis
We analyze how age groups influence car purchase amount

```python
age_bins = [20,40,50,60,70]
age_labels = ["young","middle","senior","older"]

df["age_group"] = pd.cut(df["age"], bins=age_bins, labels=age_labels)

age_analysis = df.groupby("age_group")["car purchase amount"].mean().reset_index()
age_analysis
```

```python
age_analysis.plot(x="age_group", y="car purchase amount", kind='bar')
plt.title("Car Purchase by Age Group")
plt.show()
```

*Insight:*  
Older customers tend to spend slightly more than younger customers, showing a moderate relationship.

## Country Analysis
We identify the top countries with the highest average car purchase amounts.

```python
country_analysis = df.groupby("country")["car purchase amount"].mean().reset_index()
top_country = country_analysis.sort_values(by="car purchase amount", ascending=False).head(5)
top_country
```

```python
top_country.sort_values(by="car purchase amount").plot(
    x="country", y="car purchase amount", kind='barh'
)
plt.title("Top Countries by Average Car Purchase")
plt.show()
```

*Insight:*  
Some countries have slightly higher spending, but overall differences are small.

## Credit Card Debt Analysis
We examine whether debt level affects car purchase amount

```python
credit_card_debt_bins = [0,7397,9655,11798,20000]
labels = ["low","medium","high","very_high"]

df["debt_card_group"] = pd.cut(df["credit card debt"], bins=credit_card_debt_bins, labels=labels)

credit_card_debt_analysis = df.groupby("debt_card_group")["car purchase amount"].mean().reset_index()
credit_card_debt_analysis
```

```python
credit_card_debt_analysis.plot(x="debt_card_group", y="car purchase amount", kind='bar')
plt.title("Car Purchase by Credit Card Debt")
plt.show()
```

*Insight:*  
Credit card debt shows no strong relationship with car purchase amount, as values are very similar across groups.

```python
## Conclusion

- Net worth is the strongest predictor of car purchase
- Age has moderate influence
- Gender, country, and credit card debt have weak influence

Overall, financial strength is the key driver of customer spending.
```
